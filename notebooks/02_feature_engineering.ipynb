```python
import sqlite3
import pandas as pd 
```

```python
conn = sqlite3.connect("../data/olist.sqlite")
```

```python
query = """
SELECT
    c.customer_unique_id,
    COUNT(DISTINCT o.order_id) AS total_orders,
    SUM(op.payment_value) AS total_spent,
    SUM(op.payment_value) / COUNT(DISTINCT o.order_id) AS average_order_value,
    MIN(o.order_purchase_timestamp) AS first_purchase,
    MAX(o.order_purchase_timestamp) AS last_purchase,

    julianday('2018-10-17 17:30:18')
    - julianday(MAX(o.order_purchase_timestamp))
    AS recency_days

FROM customers AS c
JOIN orders AS o
    ON c.customer_id = o.customer_id
JOIN order_payments AS op
    ON o.order_id = op.order_id

GROUP BY c.customer_unique_id

ORDER BY total_spent DESC;

"""

customer_features = pd.read_sql_query(query, conn)
```

```python
customer_features.to_csv("../data/cutomer_features.csv", index=False)
```
